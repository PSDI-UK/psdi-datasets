create functions to construct json in a semi automated fashion

```python
def list_to_cords(lst):
    '''A function which take a list of data cordinates in the form [x,y,x,y.....] and 
    converts it to a list of dictionaries in the form [{x:value,y:value},{x:value,y:value}...]
    
    Imput- list of cords of points on the graph
    Output - a list of dictionaries of datapoints'''
    data = []
    for i in range(0, len(lst), 2):
        x = lst[i]
        y = lst[i + 1]
        data.append({"x": x, "y": y})
    return data

def generate_json(keys = [
        "caption",
        "compound",
        "solvent",
        "CMC_value",
        "data",
        "line_of_best_fit",
        "units"
    ],*values,file_path):
    '''This is a function for the creation of json file with keys as defualt specified above 
    and values being a dicitionary to be linked to the keys in the function
    
    the output is saved to the file path as a json file
    
    imput-
    key- a list of keys - default is the dictionary keys we have used for our json file as defualt
    values-a list of data dictiaries
    file_path - the file path where the json file will be saved'''
    
    #iterate through the keys and values a sets them as key value 
    dictionary = {k: v for k, v in zip(keys, values)}
    
    # Save JSON to file
    with open(file_path, "w") as file:
        json.dump(dictionary, file, indent=4)
    return dictionary
    
```
